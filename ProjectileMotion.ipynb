```python
# @title
#Hyunje Kim#
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from math import pi, radians, cos, sin, exp, atan, sqrt
from IPython.display import HTML

#Spherical Object Details:
D = 0.07 #{m}
ρ = 7800 #{kg/m³}
V = (4/3)*pi*(D/2)**3
#Constants:
g = 9.8       #{m/s²}
m = ρ * V     #{kg}
γ = 0.25      #{Ns²/m⁴}
λ = 10000     #{m}
c = γ * D**2  #drag coefficient
#Initial Conditions:
x = 0  #{m}
y = 0  #{m}
v0 = 300              #{m/s}
θ = radians(50)  #{rad}
#Time {s} Parameter:
Δt = 0.01
tf = 35
#Velocity Components:
vx = v0 * cos(θ)
vy = v0 * sin(θ)
#____________________________Velocity Verlet________________________________________________
#_____________________Drag Function {c(y) = γD²e^(-y/λ)}____________
vx_cont = vx
vy_cont = vy
#pos arrays
x_cont = 0
y_cont = 0
X_cont = [x_cont]
Y_cont = [y_cont]
#drag function
c_y = c * exp(-y_cont/λ)
#acc comp.
ax_cont = -c_y*v0*vx_cont/m
ay_cont = -g -(c_y*v0*vy_cont/m)
#time parameter
t_cont = 0.01
while t_cont < tf and y_cont > -1:
    #Velocity Verlet Algorithm
    #1)New Position
    x_cont += vx_cont*t_cont + (0.5 * t_cont**2 * ax_cont)
    y_cont += vy_cont*t_cont + (0.5 * t_cont**2 * ay_cont)
    X_cont.append(x_cont)
    Y_cont.append(y_cont)
    c_y = c * exp(-y_cont/λ)
    #2)Temporary New Acceleration(old vel, new pos)
    θ1 = atan(y_cont/x_cont)               #Updating the angular component of pos & vel
    v_cont = sqrt(vx_cont ** 2 + vy_cont ** 2)  # Updating total velocity magnitude
    temp_ax_cont = -c_y*v_cont*v_cont*cos(θ1)/m
    temp_ay_cont = -g -(c_y*v_cont*v_cont*sin(θ1)/m)
    #3)Temporary New Velocity(temp acc, old acc)
    temp_vx_cont = vx_cont + ((0.5*t_cont)*(temp_ax_cont + ax_cont))
    temp_vy_cont = vy_cont + ((0.5*t_cont)*(temp_ay_cont + ay_cont))
    temp_v_cont = sqrt(temp_vx_cont**2 + temp_vy_cont**2)
    #4)New Acceleration(old vel, new pos)
    new_ax_cont = -c_y*temp_v_cont*temp_vx_cont/m
    new_ay_cont = -g -(c_y*temp_v_cont*temp_vy_cont/m)
    #5)New Velocity(new acc, old acc)
    vx_cont = vx_cont + ((0.5*t_cont)*(new_ax_cont + ax_cont))
    vy_cont = vy_cont + ((0.5*t_cont)*(new_ay_cont + ay_cont))
    #6)Acceleration Reassignment
    ax_cont = new_ax_cont
    ay_cont = new_ay_cont
    t_cont += Δt
  #______________________________Drag Coefficient {c = γD²}______________________________
#Velocity,Net Distances, & Acceleration Components
vx_c = vx
vy_c = vy
#Arrays
x_c = 0
y_c = 0
X_c = [x_c]
Y_c = [y_c]
#Acceleration Components
ax_c = -c*v0*vx_c/m
ay_c = -g -(c*v0*vy_c/m)
t_c = 0.01
while t_c < tf and y_c > -1:
    #Velocity Verlet:
    #1)New Position
    x_c += vx_c*t_c + (0.5 * t_c**2 * ax_c)
    y_c += vy_c*t_c + (0.5 * t_c**2 * ay_c)
    X_c.append(x_c)
    Y_c.append(y_c)
    #2)Temporary New Acceleration(old vel, new pos)
    θ2 = atan(y_c/x_c)              #Updating the angular component of pos & vel
    v_c = sqrt(vx_c ** 2 + vy_c ** 2)
    temp_ax_c = -c*v_c*v_c*cos(θ2)/m
    temp_ay = -g -(c*v_c*v_c*sin(θ2)/m)
    #3)Temporary New Velocity(temp acc, old acc)
    temp_vx_c = vx_c + ((0.5*t_c)*(temp_ax_c + ax_c))
    temp_vy_c = vy_c + ((0.5*t_c)*(temp_ay + ay_c))
    temp_v_c = sqrt(temp_vx_c ** 2 + temp_vy_c ** 2)
    #4)New Acceleration(old vel, new pos)
    new_ax_c = -c*temp_v_c*temp_vx_c/m
    #5)New Velocity(new acc, old acc)
    new_ay_c = -g -(c*temp_v_c*temp_vy_c/m)
    vx_c = vx_c + ((0.5*t_c)*(new_ax_c + ax_c))
    vy_c = vy_c + ((0.5*t_c)*(new_ay_c + ay_c))
    #6)Acceleration Reassignment
    ax_c = new_ax_c
    ay_c = new_ay_c
    t_c += Δt
#______________________________No Drag {c=0}______________________________
vx_vac = vx
vy_vac = vy
#Position & Arrays
x_vac = 0
y_vac = 0
X_vac = [x]
Y_vac = [y]
#time parameters
t_vac = 0.01
while t_vac < tf and y_vac > -1:
    #Velocity Verlet Algorithm
    #1)New Position
    x_vac += vx_vac * t_vac
    y_vac += vy_vac * t_vac + (0.5 * -g * t_vac**2)
    X_vac.append(x_vac)
    Y_vac.append(y_vac)
    #2)New Velocity & Increment
    vy_vac = vy_vac + (t_vac * -g)
    t_vac += Δt
#===============================================================================


# This line makes Jupyter show animations inline as HTML widgets
plt.rcParams["animation.html"] = "jshtml"

X_cont = np.array(X_cont); Y_cont = np.array(Y_cont)
X_c = np.array(X_c); Y_c = np.array(Y_c) # ignoring variable atmospheric density w.r.t. altitude
X_vac = np.array(X_vac); Y_vac = np.array(Y_vac) # ignoring both drag & variable atmospheric density w.r.t. altitude


# Set up the figure
fig, ax = plt.subplots()
ax.set_title('Y {m} vs X {m}')
ax.grid(True)
ax.set_ylabel("Y {m}"); ax.set_xlabel("X {m}")
ax.axis([0, X_vac[-1], 0, Y_vac.max()])

trajectory_continuous, = ax.plot([], [], linestyle='--', linewidth=0.7, color='red')
sphere_continuous, = ax.plot([], [], 'o', markersize=3, alpha=0.5, color='red', label=r"$\vec{F}_g + \vec{F}_{D}(y)$")

trajectory_constant, = ax.plot([], [], linestyle='--',  linewidth=0.7, color='blue')
sphere_constant, = ax.plot([], [], 'o', markersize=3, alpha=0.5, color='blue', label=r"$\vec{F}_g + \vec{F}_{D}$")

trajectory, = ax.plot([], [], linestyle='--', linewidth=0.7, color='green')
sphere, = ax.plot([], [], 'o', markersize=3, alpha=0.5, color='green', label=r"$\vec{F}_g$")

ax.legend(loc='best')

# Animate function
def animate(_):
    trajectory_continuous.set_data(X_cont[:_], Y_cont[:_])
    sphere_continuous.set_data(X_cont[_-1:_], Y_cont[_-1:_])
    trajectory_constant.set_data(X_c[:_], Y_c[:_])
    sphere_constant.set_data(X_c[_-1:_], Y_c[_-1:_])
    trajectory.set_data(X_vac[:_], Y_vac[:_])
    sphere.set_data(X_vac[_-1:_], Y_vac[_-1:_])

    # note, need to end with ',' for single returned frame
    return (trajectory_continuous, sphere_continuous,
            trajectory_constant, sphere_constant,
            trajectory, sphere)

plt.close(fig)
ani = animation.FuncAnimation(fig, animate, frames=len(X_cont), interval=80, blit=True)
HTML(ani.to_jshtml())
```
